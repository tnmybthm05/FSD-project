```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("breast-cancer.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", data.shape)
```

```python
X = data.drop(columns=["id", "diagnosis"])
y = data["diagnosis"]

# Convert diagnosis:
# M = 1 (Malignant)
# B = 0 (Benign)

y = y.map({"M": 1, "B": 0})
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```python
print("\n========================================")
print("a. SVM WITH LINEAR KERNEL")
print("========================================")

linear_model = SVC(kernel="linear")

# Train
linear_model.fit(X_train, y_train)

# Prediction
linear_pred = linear_model.predict(X_test)
 
# Accuracy
linear_accuracy = accuracy_score(y_test, linear_pred)

print("Linear Kernel Accuracy:",
      linear_accuracy * 100, "%")
```

```python
print("\n========================================")
print("b. SVM WITH POLYNOMIAL KERNEL")
print("========================================")

poly_model = SVC(kernel="poly")

# Train
poly_model.fit(X_train, y_train)

# Prediction
poly_pred = poly_model.predict(X_test)

# Accuracy
poly_accuracy = accuracy_score(y_test, poly_pred)

print("Polynomial Kernel Accuracy:",
      poly_accuracy * 100, "%")
```

```python
print("\n========================================")
print("c. SVM WITH RBF KERNEL")
print("========================================")

rbf_model = SVC(kernel="rbf")

# Train
rbf_model.fit(X_train, y_train)

# Prediction
rbf_pred = rbf_model.predict(X_test)

# Accuracy
rbf_accuracy = accuracy_score(y_test, rbf_pred)

print("RBF Kernel Accuracy:",
      rbf_accuracy * 100, "%")
```

```python
print("\n\n========================================")
print("       SVM KERNEL ACCURACY COMPARISON")
print("========================================")

print("a. Linear Kernel      :", round(linear_accuracy * 100, 2), "%")
print("b. Polynomial Kernel  :", round(poly_accuracy * 100, 2), "%")
print("c. RBF Kernel         :", round(rbf_accuracy * 100, 2), "%")
```

```python
accuracies = {
    "Linear": linear_accuracy,
    "Polynomial": poly_accuracy,
    "RBF": rbf_accuracy
}

best_kernel = max(accuracies, key=accuracies.get)

print("\nBest Performing Kernel:", best_kernel)
print("Best Accuracy:",
      round(accuracies[best_kernel] * 100, 2), "%")
```
